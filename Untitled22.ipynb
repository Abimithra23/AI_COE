```python

# Transformer-Based Next-Token Prediction
# Using Basic Self-Attention

# 1. Import libraries
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.layers import (
    Input, Embedding, Dense, LayerNormalization,
    MultiHeadAttention, Dropout, GlobalAveragePooling1D
)
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 2. Sample training text
text = """
artificial intelligence is changing the world
machine learning is a part of artificial intelligence
deep learning uses neural networks
transformers use attention to understand language
self attention helps models learn relationships between words
natural language processing helps computers understand text
artificial intelligence helps solve real world problems
machine learning models learn patterns from data
transformers are used in modern language models
deep learning is useful for image and text classification
"""

# 3. Tokenize the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

sequences = tokenizer.texts_to_sequences([text])[0]
vocab_size = len(tokenizer.word_index) + 1

print("Vocabulary Size:", vocab_size)
print("Vocabulary:", tokenizer.word_index)

# 4. Create input sequences and next-token labels
input_sequences = []

for i in range(1, len(sequences)):
    input_sequences.append(sequences[:i + 1])

max_length = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_length,
    padding="pre"
)

X = input_sequences[:, :-1]
y = input_sequences[:, -1]

sequence_length = X.shape[1]

# 5. Define a Transformer block using self-attention
inputs = Input(shape=(sequence_length,))

# Word embeddings
embedding = Embedding(
    input_dim=vocab_size,
    output_dim=32
)(inputs)

# Positional information is learned through the embedding sequence
positions = tf.range(start=0, limit=sequence_length, delta=1)
position_embedding = Embedding(
    input_dim=sequence_length,
    output_dim=32
)(positions)

x = embedding + position_embedding

# Basic self-attention with a causal mask
attention = MultiHeadAttention(
    num_heads=2,
    key_dim=16
)(
    query=x,
    value=x,
    key=x,
    use_causal_mask=True
)

# Residual connection and normalization
x = LayerNormalization()(x + attention)

# Feed-forward network
ffn = Dense(64, activation="relu")(x)
ffn = Dense(32)(ffn)

x = LayerNormalization()(x + ffn)

# Use the last token's representation for next-token prediction
x = tf.keras.layers.Lambda(lambda t: t[:, -1, :])(x)

outputs = Dense(vocab_size, activation="softmax")(x)

model = Model(inputs, outputs)

# 6. Compile the model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# 7. Train the model
history = model.fit(
    X,
    y,
    epochs=200,
    batch_size=8,
    verbose=0
)

print("\nTraining completed!")

# 8. Plot training loss
plt.figure(figsize=(7, 4))
plt.plot(history.history["loss"])
plt.title("Transformer Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()

# 9. Predict the next word
def predict_next_word(seed_text):
    for _ in range(5):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = pad_sequences(
            [token_list],
            maxlen=sequence_length,
            padding="pre",
            truncating="pre"
        )

        predictions = model.predict(token_list, verbose=0)[0]

        # Select the most likely next token
        predicted_id = int(np.argmax(predictions))

        if predicted_id == 0:
            break

        predicted_word = tokenizer.index_word.get(predicted_id)

        if predicted_word is None:
            break

        seed_text += " " + predicted_word

    return seed_text

# 10. Test the model
test_texts = [
    "artificial intelligence",
    "machine learning",
    "transformers use",
    "deep learning"
]

for seed in test_texts:
    print("\nInput:", seed)
    print("Prediction:", predict_next_word(seed))

# 11. Save the trained model
model.save("transformer_next_token.keras")
print("\nModel saved successfully!")
```
